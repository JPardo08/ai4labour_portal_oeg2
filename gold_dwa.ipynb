### Qué es “óptimo” aquí

En la Information retrieval (IR) se promedian métricas por query; la precisión estadística depende de cuántas queries tengas (no de cuántos documentos haya en el índice). Con 221 DWAs disponibles, tu límite natural es 221 queries para ambos pipelines (en DWA→Skill las queries siguen siendo DWAs; las 768 skills son los documentos, no las unidades de muestreo).

* **IR** = *Information Retrieval* → “Recuperación de Información”.
  Es el área que estudia cómo buscar y rankear documentos, pasajes, skills... a partir de queries. Todas las métricas que estás usando (P\@k, R\@k, MAP, MRR, nDCG) vienen de ahí: se evalúan **por query** y luego se hace un promedio.

Un cálculo rápido (peor caso, varianza máxima en \[0,1]) para estimar la media de una métrica como nDCG/MAP con margen de error ±e al 95%:

$$
n \;\gtrsim\; \frac{z_{0.95}^2}{4\,e^2} \quad \text{con } z_{0.95}=1{.}96
$$

* **±0.10** → **n ≈ 96** queries
* **±0.07** → **n ≈ 196** queries
* **±0.05** → **n ≈ 385** queries (no alcanzable con 221 DWAs)

Conclusiones:
* Con **todas las 221 DWAs** obtienes, en peor caso, un margen \~**±0.066** a 95% (muchísimo mejor que ±0.1).
* Con **10 queries** el margen es enorme (≈±0.31 peor caso): de ahí los “números redondos” y empates.

### Nota sobre “representatividad” en Coursera

Que haya 768 skills no cambia el tamaño muestral necesario: lo que manda es el nº de queries (DWAs). Para cubrir mejor la variabilidad de skills:

* asegúrate de que el GOLD por DWA tenga suficientes skills relevantes (si no, abundan ceros),
* y estratifica las queries para que sus relevantes cubran clusters de skills distintos (o reporta métricas por estratos y un macro-promedio).



### Recomendación práctica

* **Usa todas las 221 DWAs como queries** para ambos pipelines. Es tu “óptimo” real con el corpus actual.
* Si por coste tienes que muestrear, apunta a **≥100 queries** (te deja \~±0.10) y **estratifica** por:

  * área/IWA,
  * longitud de la DWA,
  * frecuencia/cobertura en GOLD.
    Así evitas que 2–3 queries “raras” sesguen todo.

Reporta Intervalo de confianza del 95% por sistema (bootstrap por query) y usa tests pareados (approx. randomization/boostrap) para decidir empates. Rompe empates con orden nDCG\@10 → MAP\@10 → MRR → R\@1 → coverage\_rel\@10.

* **IC** = *Intervalo de Confianza*. Es un rango estadístico que te dice en qué valores puede estar la “verdadera” media de tu métrica con cierta probabilidad (típicamente 95%).
  
Ejemplo: si con 221 queries tu `nDCG@10` medio es **0.42** y el IC95% es \[0.39, 0.45], significa que con un 95% de confianza el valor real está en ese rango.

El problema de Coursera no era la fórmula de las métricas (IR), sino que con tan pocas queries el **IC es enorme** (±0.3 con 10 queries). Eso genera resultados “redondos” y empates. Con más queries, el IC se estrecha y la leaderboard empieza a diferenciar modelos.

```python
from collections import defaultdict
from itertools import combinations
from typing import List, Dict, Any

def _norm(s: str) -> str:
    """Normaliza texto para comparar: recorta espacios y pasa a minúsculas."""
    return " ".join(str(s).split()).strip().lower()

def find_gold_duplicates(gold_list: List[Dict[str, Dict[str, List[str]]]]) -> Dict[str, Any]:
    """
    Analiza una lista de diccionarios GOLD y devuelve duplicados:
      - duplicate_queries: queries (DWAs) repetidas y dónde aparecen.
      - duplicate_skills: skills repetidas globalmente y en qué queries/buckets.
      - duplicate_skills_within_query: skills repetidas dentro de la misma query.
      - pairs_*: pares (i, j) de índices de GOLD que comparten elementos repetidos.

    Parámetro
    ---------
    gold_list: [GOLD_dict, GOLD_dict, ...]
        Cada GOLD_dict tiene la forma:
        {
          "Query A": {"positives": [...], "negatives": [...]},
          "Query B": {"positives": [...], "negatives": [...]},
          ...
        }

    Retorna
    -------
    dict con claves:
      {
        "duplicate_queries": [
          {
            "query_norm": str,
            "occurrences": [{"gold_idx": int, "query": str}],
            "pairs": [(int, int), ...]
          }, ...
        ],
        "duplicate_skills": [
          {
            "skill_norm": str,
            "occurrences": [
              {"gold_idx": int, "query": str, "bucket": "positives|negatives", "skill": str}
            ],
            "pairs": [(int, int), ...],
            "queries_involved": [str, ...]  # queries únicas donde aparece
          }, ...
        ],
        "duplicate_skills_within_query": [
          {
            "gold_idx": int,
            "query": str,
            "duplicates": [
              {"skill_norm": str, "originals": [str, ...], "buckets": ["positives|negatives", ...]}
            ]
          }, ...
        ]
      }
    """
    # --- Índices de aparición ---
    query_locs = defaultdict(list)   # q_norm -> [{"gold_idx", "query"}...]
    skill_locs = defaultdict(list)   # s_norm -> [{"gold_idx", "query", "bucket", "skill"}...]

    # --- Duplicados dentro de una misma query ---
    dup_within_query = []

    for gold_idx, gold in enumerate(gold_list):
        if not isinstance(gold, dict):
            continue  # ignora entradas que no sean diccionarios

        for query, payload in gold.items():
            q_norm = _norm(query)
            query_locs[q_norm].append({"gold_idx": gold_idx, "query": query})

            pos = list(map(str, payload.get("positives", []) or []))
            neg = list(map(str, payload.get("negatives", []) or []))

            # Registra localizaciones globales por skill
            for s in pos:
                skill_locs[_norm(s)].append({
                    "gold_idx": gold_idx, "query": query, "bucket": "positives", "skill": s
                })
            for s in neg:
                skill_locs[_norm(s)].append({
                    "gold_idx": gold_idx, "query": query, "bucket": "negatives", "skill": s
                })

            # Detecta duplicados dentro de la misma query (positives, negatives o cruzados)
            bucket_tagged = [("positives", s) for s in pos] + [("negatives", s) for s in neg]
            counts = defaultdict(lambda: {"originals": [], "buckets": []})
            for b, s in bucket_tagged:
                ns = _norm(s)
                counts[ns]["originals"].append(s)
                counts[ns]["buckets"].append(b)

            dups_here = []
            for ns, info in counts.items():
                if len(info["originals"]) > 1:
                    dups_here.append({
                        "skill_norm": ns,
                        "originals": info["originals"],
                        "buckets": info["buckets"]
                    })
            if dups_here:
                dup_within_query.append({
                    "gold_idx": gold_idx,
                    "query": query,
                    "duplicates": dups_here
                })

    # --- Queries repetidas entre GOLDs ---
    duplicate_queries = []
    for qn, locs in query_locs.items():
        if len(locs) > 1:
            # pares de índices (i, j) donde aparece
            idxs = sorted({loc["gold_idx"] for loc in locs})
            pairs = sorted(set(combinations(idxs, 2)))
            duplicate_queries.append({
                "query_norm": qn,
                "occurrences": locs,
                "pairs": pairs
            })

    # --- Skills repetidas globalmente entre GOLDs (y/o entre queries) ---
    duplicate_skills = []
    for sn, locs in skill_locs.items():
        if len(locs) > 1:
            idxs = sorted({loc["gold_idx"] for loc in locs})
            pairs = sorted(set(combinations(idxs, 2)))
            queries_involved = sorted({loc["query"] for loc in locs})
            duplicate_skills.append({
                "skill_norm": sn,
                "occurrences": locs,
                "pairs": pairs,
                "queries_involved": queries_involved
            })

    return {
        "duplicate_queries": duplicate_queries,
        "duplicate_skills": duplicate_skills,
        "duplicate_skills_within_query": dup_within_query
    }


def pretty_print_duplicates(report: Dict[str, Any]) -> None:
    """Imprime un resumen legible del informe de duplicados."""
    dq = report.get("duplicate_queries", [])
    ds = report.get("duplicate_skills", [])
    dw = report.get("duplicate_skills_within_query", [])

    print("=== QUERIES REPETIDAS ENTRE DICCIONARIOS ===")
    if not dq:
        print("✔ No hay queries repetidas.")
    else:
        for item in dq:
            qs = ", ".join(sorted({occ["query"] for occ in item["occurrences"]}))
            idxs = sorted({occ["gold_idx"] for occ in item["occurrences"]})
            print(f"- Query: {qs}")
            print(f"  · DICCIONARIOS: {idxs}")
            print(f"  · PARES: {item['pairs']}")

    print("\n=== SKILLS REPETIDAS ENTRE DICCIONARIOS/QUERIES ===")
    if not ds:
        print("✔ No hay skills repetidas.")
    else:
        for item in ds:
            skills_orig = sorted({occ["skill"] for occ in item["occurrences"]})
            idxs = sorted({occ["gold_idx"] for occ in item["occurrences"]})
            queries = item["queries_involved"]
            print(f"- Skill(s): {skills_orig}")
            print(f"  · DICCIONARIOS: {idxs}")
            print(f"  · PARES: {item['pairs']}")
            print(f"  · QUERIES: {queries}")

    print("\n=== SKILLS DUPLICADAS DENTRO DE LA MISMA QUERY ===")
    if not dw:
        print("✔ No hay duplicados internos dentro de las queries.")
    else:
        for item in dw:
            print(f"- GOLD #{item['gold_idx']} · Query: {item['query']}")
            for d in item["duplicates"]:
                print(f"  · Skill_norm: {d['skill_norm']}")
                print(f"    Originals: {d['originals']}")
                print(f"    Buckets: {d['buckets']}")

    
```

```python
gold_list = [GOLD, GOLD2]

report = find_gold_duplicates(gold_list)
pretty_print_duplicates(report)

# Si prefieres inspeccionarlo como dict:
# from pprint import pprint
# pprint(report)
```

```python
def merge_gold_dicts(gold_dicts):
    """
    Recibe una lista de diccionarios GOLD y devuelve uno solo con todos los pares.
    Asume que las queries (claves) ya están normalizadas y no se repiten entre diccionarios.
    """
    merged = {}
    for gold in gold_dicts:
        for query, pairs in gold.items():
            if query in merged:
                raise ValueError(f"Query duplicada encontrada: {query}")
            merged[query] = pairs
    return merged
```
